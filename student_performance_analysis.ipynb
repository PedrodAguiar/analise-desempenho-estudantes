**Importando as Bibliotecas**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

# **Importando e entendendo a base de dados**

```python
# Lendo dados
df = pd.read_csv("/content/drive/MyDrive/dados_csv/student_habits_performance.csv")
```

```python
# visualizando dados
df
```

# **Quais hábitos imapactam mais o desempenho dos alunos**

```python
# Tipos de dados
df.info()
```

```python
# Colunas numéricas
cols = [
    "study_hours_per_day",
    "social_media_hours",
    "netflix_hours",
    "sleep_hours",
    "attendance_percentage",
    "exercise_frequency",
    "mental_health_rating",
    "exam_score",
]

# Plotar mapa de calor (heatmap)
sns.heatmap(df[cols].corr(),annot=True,cmap="coolwarm")
plt.title("Correlação entre hábitos e nota final")
plt.show()
```

## **Os Alunos que mais estudam tem os melhores desempenhos?**

```python
# Gráfico de dispersão com linha de regressão
# x="study_hours_per_day" / y="exam_score"
sns.lmplot(data=df, x="study_hours_per_day", y="exam_score")
plt.title("Mais estudo + notas mais altas?")
plt.show()
```

```python
# Comparando médias: quem estuda >5h X <2h
filtro_estudo_alto = df['study_hours_per_day'] > 5
filtro_estudo_baixo = df['study_hours_per_day'] < 2

grupo_estudo_alto = df[filtro_estudo_alto]["exam_score"]
grupo_estudo_baixo = df[filtro_estudo_baixo]["exam_score"]

print("Média notas (estuda >5h):", grupo_estudo_alto.mean())
print("Média notas (estuda <2h):", grupo_estudo_baixo.mean())
```

## **O tempo gasto nas redes sociais afeta o desempenho dos alunos?**

```python
# Redes sociais: distribuição geral (Histograma)
# x="social_media_hours"
sns.histplot(data=df, x="social_media_hours")
```

```python
from IPython.core.interactiveshell import softspace
# Avaliando notas médias
# por diferentes intervalos (bins) de períodos gastos em redes sociais
# ["0-2h", "2h-4h", "4h-6h", "6h+"]
df["social_media_bin"] = pd.cut(
    df['social_media_hours'],
    bins=[0, 2, 4, 6, 8],
    labels=["0-2h", "2h-4h", "4h-6h", "6h+"]
)


# Plotar gráfico de caixa (boxplot)
sns.boxplot(data=df,x="social_media_bin",y="exam_score")
plt.title("Notas por tempo em redes sociais")
plt.show()
```

## **Alunos mais saudáveis têm melhores desempenhos?**

```python
# Frequência de exercícios físicos
for col in ["exercise_frequency", "mental_health_rating", "diet_quality"]:
  sns.boxplot(x=col, y="exam_score", data = df)
  plt.show()
```

## **Há diferença nas notas entre homens e mulheres?**

```python
# Estatística entre gênero (média e descvio padão)
df.groupby(["gender"])["exam_score"].agg(["mean","std"])
```

```python
#Validar distribuição de gênero
df["gender"].value_counts(normalize=True)
```

## **A frequência nas aulas influencia a nota?**

```python
sns.scatterplot(data=df, x='attendance_percentage', y="exam_score")
plt.title("Correlação entre frequência e nota final")
plt.show()
```
